```python
import openrouteservice
import numpy as np
import time


client = openrouteservice.Client(key=API_KEY)

def get_batched_distance_matrix(coords, chunk_size=50):
    n = len(coords)
    
    full_matrix = np.zeros((n, n))
    
    print(f"Total points: {n}. Total routes to calculate: {n * n}")
    print("Fetching batches. This may take a minute...")

   
    for i in range(0, n, chunk_size):
        for j in range(0, n, chunk_size):

            row_chunk = coords[i:i + chunk_size]
            col_chunk = coords[j:j + chunk_size]
            
         
            locations = row_chunk + col_chunk
            

            sources = list(range(len(row_chunk)))
            destinations = list(range(len(row_chunk), len(row_chunk) + len(col_chunk)))
            
            try:

                response = client.distance_matrix(
                    locations=locations,
                    sources=sources,
                    destinations=destinations,
                    profile='cycling-regular',
                    metrics=['distance'],
                    units='km'
                )
                

                sub_matrix = np.array(response['distances'])
                full_matrix[i:i + chunk_size, j:j + chunk_size] = sub_matrix
                
                print(f"Successfully fetched block [{i}:{i+chunk_size}, {j}:{j+chunk_size}]")
                
               
                time.sleep(1.5)
                
            except Exception as e:
                print(f"Error fetching block [{i}:{i+chunk_size}, {j}:{j+chunk_size}]: {e}")

                return full_matrix
                
    return full_matrix
final_matrix = get_batched_distance_matrix(coordinates)

print("\nFinal Complete Matrix:")
print(np.round(final_matrix, 2))
```
